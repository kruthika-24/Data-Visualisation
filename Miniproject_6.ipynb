```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("zomato.csv")
df.head()
```

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
df.describe()
```

```python
df = df.drop_duplicates()

df['rate_num'] = df['rate'].astype(str).str.replace('/5', '', regex=False).str.strip()
df['rate_num'] = pd.to_numeric(df['rate_num'], errors='coerce')

df['approx_cost'] = df['approx_cost(for two people)'].astype(str).str.replace(',', '')
df['approx_cost'] = pd.to_numeric(df['approx_cost'], errors='coerce')

df.loc[(df['rate_num'] < 0) | (df['rate_num'] > 5), 'rate_num'] = np.nan

df['rate_num'] = df['rate_num'].fillna(df['rate_num'].median())
df['approx_cost'] = df['approx_cost'].fillna(df['approx_cost'].median())

for col in ['online_order', 'book_table', 'listed_in(type)']:
    df[col] = df[col].fillna(df[col].mode()[0])

print(df.isnull().sum())
```

```python
df.head()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x='listed_in(type)',
    order=df['listed_in(type)'].value_counts().index
)

plt.title('Number of Restaurants in Each Restaurant Type')
plt.xlabel('Restaurant Type')
plt.ylabel('Number of Restaurants')
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='rate_num',
    y='votes'
)

plt.title('Votes vs Restaurant Ratings')
plt.xlabel('Restaurant Rating')
plt.ylabel('Votes')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='rate_num',
    bins=10,
    kde=True
)

plt.title('Distribution of Restaurant Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Restaurants')
plt.show()
```

```python
plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x='rate_num',
    y='votes',
    size='approx_cost',
    hue='listed_in(type)',
    sizes=(30,300),
    alpha=0.7
)

plt.title('Rating vs Votes with Approximate Cost')
plt.xlabel('Restaurant Rating')
plt.ylabel('Votes')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='online_order',
    y='rate_num'
)

plt.title('Restaurant Ratings Based on Online Ordering')
plt.xlabel('Online Ordering')
plt.ylabel('Rating')
plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x='online_order'
)

plt.title('Online Ordering Availability')
plt.xlabel('Online Ordering')
plt.ylabel('Number of Restaurants')
plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x='book_table'
)

plt.title('Table Booking Availability')
plt.xlabel('Table Booking')
plt.ylabel('Number of Restaurants')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='approx_cost',
    bins=10,
    kde=True
)

plt.title('Distribution of Approximate Cost for Two People')
plt.xlabel('Approximate Cost')
plt.ylabel('Number of Restaurants')
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    data=df,
    x='listed_in(type)',
    y='rate_num'
)

plt.title('Restaurant Ratings by Restaurant Type')
plt.xlabel('Restaurant Type')
plt.ylabel('Rating')
plt.xticks(rotation=30)
plt.show()
```

```python
top_restaurants = (
    df.groupby('name')['votes']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,6))

top_restaurants.sort_values().plot(kind='barh')

plt.title('Top 10 Restaurants by Votes')
plt.xlabel('Total Votes')
plt.ylabel('Restaurant')
plt.show()
```

```python
correlation = df[['rate_num', 'votes', 'approx_cost']].corr()

print(correlation)
```

```python
plt.figure(figsize=(7,5))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Between Rating, Votes and Approximate Cost')
plt.show()
```
